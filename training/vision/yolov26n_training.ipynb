```python
import os
from collections import Counter

label_folder = os.path.join(dataset_path, "Data_YOLO/labels/train")

class_counts = Counter()

files = [
    f for f in os.listdir(label_folder)
    if f.endswith(".txt")
][:100]

for filename in files:
    with open(os.path.join(label_folder, filename), "r") as f:
        for line in f:
            values = line.split()
            if values:
                class_counts[int(values[0])] += 1

print("YOLO class IDs:")
print(sorted(class_counts.keys()))
```

```python
from ultralytics import YOLO

model = YOLO("yolo26n.pt")

print("YOLO26n model loaded successfully!")
```

```python
results = model.train(
    data=yaml_path,
    epochs=50,
    imgsz=640,
    batch=8,
    optimizer="AdamW",
    lr0=0.001,
    patience=10,
    device=0,
    workers=2,
    cache=False,
    project="/kaggle/working/Results",
    name="PCB_YOLO26n",
    save=True,
    plots=True
)
```

```python
metrics = model.val(
    data=yaml_path,
    split="val",
    imgsz=640,
    batch=8,
    device=0
)

print("Validation completed!")
print("mAP50:", metrics.box.map50)
print("mAP50-95:", metrics.box.map)
print("Precision:", metrics.box.mp)
print("Recall:", metrics.box.mr)
```
